1.What is the difference between "Love" and "love" in NLP?

Ans:

In NLP, "Love" and "love" are treated as different words because they are case-sensitive.


"Love"(Uppercase L) - may appear at the beginning of a sentence or as a proper noun.


"love"(lowercas l ) - used as a regular word verb or noun.


Case differences can create duplicate features and affect model accuracy,so it is converted to lower case.

4.Explain the difference between stemming and lemmatization.

Ans:


Stemming: Reduces words to their root form by cutting off parts Does not always produce valid words

Examples:

"running" → "run" "studies" → "studi" (not a proper word)

Lemmatization: Converts words to their correct base or dictionary form Uses grammar and vocabulary

Examples:

"running" → "run" "studies" → "study"


Task 2:Building Advanced Preprocessing Function



```python
import re

def preprocess_text(text):
    # convert everything to lowercase
    text = text.lower()

    # remove links (http / https / www)
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # remove email addresses
    text = re.sub(r'\S+@\S+', '', text)

    # remove digits
    text = re.sub(r'[0-9]+', '', text)

    # reduce repeated characters (like soooo → so)
    text = re.sub(r'(.)\1{2,}', r'\1', text)

    # remove special characters
    text = re.sub(r'[^\w\s]', '', text)

    # remove extra spaces
    text = ' '.join(text.split())

    # filter short words but keep "no" and "not"
    words = text.split()
    cleaned_words = []

    for w in words:
        if len(w) > 2 or w in ["no", "not"]:
            cleaned_words.append(w)

    return " ".join(cleaned_words)


# taking input from user
user_text = input("Enter any sentence: ")

# processing
result = preprocess_text(user_text)

# output
print("\nCleaned Text:")
print(result)
```

Task 4: Token Analytics

```python
import re

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)
    text = re.sub(r'[0-9]+', '', text)
    text = re.sub(r'(.)\1{2,}', r'\1', text)
    text = re.sub(r'[^\w\s]', '', text)
    text = ' '.join(text.split())

    words = text.split()
    cleaned_words = []

    for w in words:
        if len(w) > 2 or w in ["no", "not"]:
            cleaned_words.append(w)

    return cleaned_words


def token_analysis(tokens):
    total = len(tokens)
    unique = len(set(tokens))

    if total == 0:
        avg_len = 0
    else:
        avg_len = sum(len(w) for w in tokens) / total

    return total, unique, round(avg_len, 2)


# Test sentences
sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

# Run analysis
for i, s in enumerate(sentences, 1):
    tokens = preprocess_text(s)
    total, unique, avg_len = token_analysis(tokens)

    print(f"\nSentence {i}")
    print("Tokens:", tokens)
    print("Total Tokens:", total)
    print("Unique Tokens:", unique)
    print("Average Token Length:", avg_len)
```

Task 5: Frequency Analysis


```python
import re
from collections import Counter

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)
    text = re.sub(r'[0-9]+', '', text)
    text = re.sub(r'(.)\1{2,}', r'\1', text)
    text = re.sub(r'[^\w\s]', '', text)
    text = ' '.join(text.split())

    words = text.split()
    cleaned_words = []

    for w in words:
        if len(w) > 2 or w in ["no", "not"]:
            cleaned_words.append(w)

    return cleaned_words


# input sentences
sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

#  combine all tokens
all_tokens = []

for s in sentences:
    tokens = preprocess_text(s)
    all_tokens.extend(tokens)

#  count frequency
freq = Counter(all_tokens)

#  top 10 most frequent words
top_10 = freq.most_common(10)

#  top 5 least frequent words
least_5 = sorted(freq.items(), key=lambda x: x[1])[:5]

#  output
print("Top 10 Most Frequent Words:")
for word, count in top_10:
    print(word, ":", count)

print("\nTop 5 Least Frequent Words:")
for word, count in least_5:
    print(word, ":", count)
```

6.Full Pipeline

```python
import re

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)
    text = re.sub(r'[0-9]+', '', text)
    text = re.sub(r'(.)\1{2,}', r'\1', text)
    text = re.sub(r'[^\w\s]', '', text)
    text = ' '.join(text.split())

    words = text.split()
    cleaned_words = []

    for w in words:
        if len(w) > 2 or w in ["no", "not"]:
            cleaned_words.append(w)

    return cleaned_words


def pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens = preprocess_text(text)
        clean_sentence = " ".join(tokens)

        all_tokens.extend(tokens)
        clean_sentences.append(clean_sentence)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }


# 🔹 Example usage
sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Visit https://openai.com now!"
]

result = pipeline(sentences)

print("All Tokens:")
print(result["tokens"])

print("\nClean Sentences:")
for s in result["clean_sentences"]:
    print(s)
```

```python
import re

def preprocess_text(text):
    # handle empty or None input
    if not text or text.strip() == "":
        return []

    text = text.lower()

    # remove urls and emails
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)

    # remove numbers
    text = re.sub(r'[0-9]+', '', text)

    # handle repeated characters
    text = re.sub(r'(.)\1{2,}', r'\1', text)

    # remove special characters (including emojis)
    text = re.sub(r'[^\w\s]', '', text)

    # remove extra spaces
    text = ' '.join(text.split())

    # split into words
    words = text.split()
    cleaned_words = []

    for w in words:
        if len(w) > 2 or w in ["no", "not"]:
            cleaned_words.append(w)

    return cleaned_words


def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens = preprocess_text(text)

        # handle cases where nothing is left after cleaning
        if not tokens:
            clean_sentence = ""
        else:
            clean_sentence = " ".join(tokens)

        all_tokens.extend(tokens)
        clean_sentences.append(clean_sentence)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }


#  Test Edge Cases
test_inputs = [
    "",                 # empty string
    "😍😍😂😂🔥🔥",       # only emojis
    "1234567890",       # only numbers
    "Normal sentence here"
]

result = full_pipeline(test_inputs)

print("Tokens:", result["tokens"])
print("Clean Sentences:", result["clean_sentences"])
```
